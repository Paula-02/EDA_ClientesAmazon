```python
!pip install kagglehub
import kagglehub
import os
import shutil
from pathlib import Path

# Download latest version
path = kagglehub.dataset_download("swathiunnikrishnan/amazon-consumer-behaviour-dataset")

print("Path to dataset files:", path)
```

```python
# Ruta de origen (la que imprime kagglehub)
src_dir = Path(r"C:\Users\anaco\.cache\kagglehub\datasets\swathiunnikrishnan\amazon-consumer-behaviour-dataset\versions\1")

# Ruta de destino
dst_dir = Path("src/data")

# Crear carpeta destino si no existe
dst_dir.mkdir(parents=True, exist_ok=True)

# Copiar todos los archivos (y subcarpetas) del dataset al repo
for item in src_dir.iterdir():
    target = dst_dir / item.name
    if item.is_dir():
        shutil.copytree(item, target, dirs_exist_ok=True)
    else:
        shutil.copy2(item, target)

print("Dataset copiado a", dst_dir.resolve())
```
